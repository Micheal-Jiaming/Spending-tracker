```python
import spending as sp

data = sp.load_spending()
print(data)
print(sp.category_total(data, "Groceries", "2026-10"))
print(sp.top_category(data, "2026-10"))
print(round(sp.month_total(data, "2026-10"), 2)) #Keep 2 decimal places
```

```python
#Self-check 1 : The data is not empty
assert len(data) > 0, "No records loaded"

#Self-check 2: The answer I calculated manually
assert round(sp.category_total(data, "Groceries", "2026-10"), 2) == 118.83, "The value for October's groceries should be 118.83."

print("All self-checks passed.")
```

```python
#Question 1:Which category did most of your spending go toward this month?
top = sp.top_category(data, "2026-10")
amount = round(sp.category_total(data, top, "2026-10"), 2)
print("In October, the most of my money went to", top, "with", amount)
```

```python
#Question 2:Which do you spend more on: groceries or dining out?
groceries = round(sp.category_total(data, "Groceries", "2026-10"), 2)
dining = round(sp.category_total(data, "Dining", "2026-10"), 2)
if groceries > dining:
    print("The category with the most spending in October was groceries, it took", groceries)
else:
    print("The category with the most spending in October was dining, it took", dining)
```

```python
#Question 3:Comparing October and November, in which month did you spend more?
october = round(sp.month_total(data, "2026-10"), 2)
november = round(sp.month_total(data, "2026-9"), 2)
if october > november:
    print("The amount spent in October was higher, at", october)
else:
    print("The amount spent in November was higher, at", november)
```
